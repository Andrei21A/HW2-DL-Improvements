```python
# !!! CHANGE THIS ONLY IF YOU WANT TO BE SCORED BASED ON THE [Custom VGG] EVALUATION CRITERIA !!!

# Random list of numbers [83, 65, 42, 55, 34, 4, 45, 18, 26, 17, 51, 91, 10, 99, 3, 98, 10, 42, 5, 75, 81, 33, 14, 59, 65, 61, 17, 30, 13, 90, 31, 49, 64, 28, 74, 57, 90, 81, 34, 99, 54, 91, 0, 19, 65, 21, 80, 28, 84, 34, 74, 60, 73, 58, 48, 55, 52, 70, 74, 25, 29, 41, 31, 9, 1, 35, 56, 91, 12, 14, 1, 71, 36, 27, 42, 45, 44, 17, 14, 39, 99, 77, 20, 93, 85, 60, 42, 23, 90, 54, 97, 9, 66, 88, 51, 54, 33, 66, 1, 70, 1, 92, 46, 15, 43, 31, 64, 1, 28, 60, 95, 17, 84, 28, 5, 49, 92, 1, 62, 65, 38, 14, 22, 79, 80, 24, 77, 35, 3, 91, 34, 42, 42, 12, 1, 39, 61, 67, 17, 89, 5, 48, 43, 10, 11, 20, 63, 41, 97, 21, 25, 48, 31, 88, 76, 45, 76, 38, 33, 37, 37, 30, 24, 59, 50, 86, 58, 71, 33, 18, 38, 15, 70, 16, 71, 40, 73, 26, 30, 58, 52, 20, 81, 56, 5, 47, 12, 77, 48, 21, 14, 10, 9, 42, 69, 25, 3, 53, 82, 97, 77, 80, 7, 2, 96, 13, 99, 58, 1, 34, 5, 50, 92, 58, 91, 27, 16, 73, 90, 11, 56, 93, 78, 21, 75, 38, 57, 8, 76, 54, 42, 24, 69, 51, 43, 17, 82, 62, 86, 23, 24, 12, 37, 91, 42, 60, 96, 93, 83, 81, 46, 13, 84, 67, 62, 78, 3, 54, 17, 48, 82, 67, 44, 16, 61, 3, 24, 61, 71, 90, 46, 40, 41, 81, 97, 97, 68, 23, 77, 40, 20, 75, 86, 10, 81, 8, 55, 56, 66, 31, 7, 82, 74, 92, 51, 84, 80, 21, 0, 0, 8, 26, 23, 8, 43, 4, 0, 63, 47, 13, 14, 90, 21, 60, 94, 95, 7, 25, 43, 88, 62, 59, 18, 81, 99, 99, 10, 64, 9, 32, 22, 91, 35, 27, 37, 29, 57, 20, 28, 42, 38, 19, 11, 11, 59, 56, 27, 55, 23, 49, 77, 31, 91, 68, 50, 71, 99, 45, 77, 82, 45, 45, 66, 61, 46, 67, 75, 66, 77, 38, 30, 62, 75, 50, 71, 17, 99, 42, 66, 79, 63, 79, 46, 93, 54, 42, 45, 65, 16, 5, 33, 82, 70, 53, 18, 34, 0, 30, 45, 2, 5, 57, 99, 33, 22, 44, 85, 12, 73, 34, 53, 42, 9, 12, 70, 20, 24, 0, 30, 30, 9, 15, 92, 22, 43, 8, 50, 98, 55, 18, 50, 78, 17, 32, 81, 61, 62, 66, 54, 50, 8, 21, 96, 80, 20, 70, 93, 49, 41, 43, 49, 44, 71, 37, 28, 30, 36, 25, 49, 70, 70, 29, 27, 46, 96, 58, 28, 73, 53, 82, 49, 18, 8, 25, 29, 34, 36, 45, 86, 3, 45, 46, 39, 26, 57, 59, 76, 8, 90, 45, 90, 78, 23, 19, 8, 37, 31, 51, 93, 68, 31, 84, 95, 53, 45, 56, 31, 54, 92, 33, 97, 45, 12, 89, 77, 60, 85, 4, 87, 86, 53, 81, 97, 73, 47, 4, 22, 99, 54, 23, 57, 81, 36, 55, 61, 42, 48, 88, 71, 27, 65, 47, 57, 40, 2, 36, 50, 78, 94, 53, 43, 72, 19, 62, 37, 8, 52, 10, 2, 25, 84, 29, 23, 85, 48, 6, 68, 5, 55, 32, 4, 4, 68, 47, 46, 31, 28, 1, 89, 13, 77, 24, 96, 33, 94, 69, 47, 98, 81, 62, 67, 59, 10, 87, 92, 19, 89, 78, 56, 78, 37, 39, 65, 74, 4, 5, 41, 26, 36, 4, 26, 76, 24, 68, 8, 5, 44, 16, 51, 87, 41, 42, 32, 5, 84, 49, 53, 52, 9, 58, 32, 31, 89, 19, 42, 7, 54, 40, 69, 10, 63, 91, 99, 15, 52, 29, 98, 93, 84, 61, 50, 15, 12, 28, 82, 27, 1, 16, 62, 93, 63, 76, 33, 52, 72, 68, 87, 49, 34, 15, 12, 1, 68, 94, 39, 76, 14, 20, 50, 62, 63, 70, 15, 35, 87, 37, 3, 96, 37, 74, 67, 70, 73, 68, 28, 40, 39, 9, 38, 19, 55, 63, 39, 15, 40, 6, 42, 84, 34, 11, 32, 17, 84, 68, 50, 65, 43, 65, 34, 85, 71, 75, 28, 52, 2, 58, 57, 61, 80, 14, 71, 26, 99, 38, 81, 95, 98, 89, 71, 75, 78, 95, 55, 51, 4, 52, 18, 18, 24, 9, 82, 3, 42, 38, 62, 88, 10, 86, 48, 12, 36, 40, 49, 81, 71, 81, 71, 67, 58, 45, 66, 24, 92, 0, 80, 90, 95, 36, 10, 94, 38, 70, 46, 52, 0, 96, 45, 56, 31, 94, 52, 97, 13, 68, 73, 65, 12, 57, 38, 87, 74, 62, 69, 79, 33, 78, 94, 61, 97, 94, 78, 57, 19, 83, 17, 96, 31, 91, 26, 81, 46, 44, 57, 67, 16, 79, 54, 4, 4, 55, 52, 57, 64, 51, 82, 12, 49, 94, 53, 72, 51, 24, 2, 70, 56, 36, 56, 35, 31, 83, 71, 11, 87, 97, 8, 80, 14, 24, 65, 74, 19, 7, 37, 46, 54, 63, 74, 21, 93, 32, 37, 24, 72, 41, 97, 37, 0, 98, 45, 9, 85, 41, 97, 96, 0, 57, 25, 39, 65, 11, 32, 63, 96, 94, 25, 76, 54, 20, 57, 75, 93, 35, 76, 10, 12, 97, 44, 79, 60, 34, 52, 11, 72, 10, 56, 85, 52, 83, 15, 84, 65, 50, 94, 89, 24, 19, 37, 66, 81, 81, 41, 38, 60, 62, 66, 98, 89, 47, 97, 45, 54, 84, 13, 6, 65, 12, 95, 78, 93, 67, 32, 29, 11, 89, 68, 34, 7, 79, 54, 51, 15, 2, 34, 56, 47, 54, 67, 56, 11, 94, 39, 73, 10, 28, 31, 46, 81, 1, 2, 5, 27, 62, 45, 80, 83, 90, 9, 5, 46, 45, 65, 73, 14, 74, 87, 31, 34, 59, 77, 50]

def create_block(in_channels: int, out_channels: int):
    # If we remove the bias from the convolution, would it be faster?
    # Or will it just be less biased?
    return [
        # When will bias stop having an impact? Correct answers might receive bonus points.
        nn.Conv2d(in_channels=in_channels, out_channels=out_channels, kernel_size=3, padding=1),
        nn.BatchNorm2d(out_channels),
        nn.ReLU(inplace=True),  # Inplace operations are faster
        nn.Identity(),  # Useless operations are slower
        nn.Conv2d(in_channels=out_channels, out_channels=out_channels, kernel_size=3, padding=1),
        # Depthwise convolutions are more efficient, parameter-wise
        # Maybe an ASPP block will bring an unexpected surprise here
        # Updating the final layers of VGG-13, especially the classification head, is another great avenue for improving the performance
        nn.BatchNorm2d(out_channels),
        *([nn.Identity()] if use_another_useless_operation else []),
        nn.ReLU(inplace=True),
        nn.MaxPool2d(kernel_size=2, stride=2)
    ]
    # Returning a tuple is faster

# !!! CHANGE THIS ONLY IF YOU WANT TO BE SCORED BASED ON THE [Custom VGG] EVALUATION CRITERIA !!!
```

```python
_per_dev = {}
def _consts(dev):
    if dev not in _per_dev:
        _per_dev[dev] = dict(mean=MEAN.to(dev), std=STD.to(dev),
                             gray=torch.tensor([0.299, 0.587, 0.114], device=dev).view(1, 3, 1, 1),
                             smooth=(torch.tensor([[1., 1., 1.], [1., 5., 1.], [1., 1., 1.]], device=dev) / 13)
                                    .expand(3, 1, 3, 3).contiguous())
    return _per_dev[dev]

def normalize(x: Tensor) -> Tensor:
    c = _consts(x.device)
    return (x - c["mean"]) / c["std"]

def to_unit(xb_u8: Tensor) -> Tensor:          
    x = xb_u8.float().div_(255.0)
    if INPUT_SIZE != x.shape[-1]:
        x = F.interpolate(x, size=(INPUT_SIZE, INPUT_SIZE), mode="bilinear", align_corners=False)
    return x

def prep(xb_u8: Tensor) -> Tensor:             
    return normalize(to_unit(xb_u8))

def _blend(img, degenerate, f):                
    return (degenerate + f * (img - degenerate)).clamp(0, 1)

def color_aug(x: Tensor, mag: float = 0.8) -> Tensor:
    # One random photometric op per image with random strength (TrivialAugment-style), vectorised.
    # Ops: identity, brightness, contrast, saturation, sharpness, solarize, posterize, autocontrast, equalize.
    B, C, H, W = x.shape
    c = _consts(x.device)
    op = torch.randint(0, 9, (B, 1, 1, 1), device=x.device)
    u = torch.rand(B, 1, 1, 1, device=x.device)
    f = 1 + (2 * u - 1) * mag                                      # enhance factor in [1-mag, 1+mag]
    gray = (x * c["gray"]).sum(1, keepdim=True)
    blur = F.conv2d(F.pad(x, (1, 1, 1, 1), mode="replicate"), c["smooth"], groups=C)
    v = (x * 255).round()
    q = torch.exp2(6 - (u * 7).floor().clamp(max=6))              # posterize: keep 2..8 bits
    lo, hi = x.amin((2, 3), keepdim=True), x.amax((2, 3), keepdim=True)
    flat = v.reshape(B * C, H * W)
    srt = flat.sort(1).values
    cdf = torch.searchsorted(srt, flat, right=True).float()               # pixels <= value
    cmin = torch.searchsorted(srt, srt[:, :1].contiguous(), right=True).float()
    span = H * W - cmin
    eq = torch.where(span > 0, (cdf - cmin) / span.clamp(min=1), flat / 255).view(B, C, H, W)
    cands = [
        x,
        (x * f).clamp(0, 1),
        _blend(x, gray.mean((2, 3), keepdim=True), f),
        _blend(x, gray, f),
        _blend(x, blur, f),
        torch.where(x >= u, 1 - x, x),
        (v / q).floor() * q / 255,
        torch.where(hi > lo, (x - lo) / (hi - lo).clamp(min=1e-6), x),
        eq,
    ]
    out = x
    for i in range(1, len(cands)):
        out = torch.where(op == i, cands[i], out)
    return out

def augment(x: Tensor, max_rot=10, max_scale=0.10, max_shift=0.10, max_shear=0.10) -> Tensor:
    B = x.size(0)
    r = lambda m: (torch.rand(B, device=x.device) * 2 - 1) * m
    ang, sc = r(math.radians(max_rot)), 1 + r(max_scale)
    cos, sin = torch.cos(ang) / sc, torch.sin(ang) / sc
    theta = torch.zeros(B, 2, 3, device=x.device)
    theta[:, 0, 0], theta[:, 0, 1], theta[:, 0, 2] = cos, -sin + r(max_shear), r(max_shift)
    theta[:, 1, 0], theta[:, 1, 1], theta[:, 1, 2] = sin, cos, r(max_shift)
    grid = F.affine_grid(theta, list(x.shape), align_corners=False)
    return F.grid_sample(x, grid, padding_mode="border", align_corners=False)

def cutmix(x, y, alpha=1.0):
    lam = np.random.beta(alpha, alpha)
    idx = torch.randperm(x.size(0), device=x.device)
    H = W = x.size(2)
    r = math.sqrt(1 - lam)
    ch, cw = int(H * r), int(W * r)
    cy, cx = np.random.randint(H), np.random.randint(W)
    y1, y2 = max(cy - ch // 2, 0), min(cy + ch // 2, H)
    x1, x2 = max(cx - cw // 2, 0), min(cx + cw // 2, W)
    x[:, :, y1:y2, x1:x2] = x[idx, :, y1:y2, x1:x2]
    lam = 1 - (y2 - y1) * (x2 - x1) / (H * W)
    return x, y, y[idx], lam

def make_batch(idx: Tensor, strong: bool):
    x, y = to_unit(Xtr[idx]), Ytr[idx]
    if USE_FLIP:
        flip = torch.rand(x.size(0), device=x.device) < 0.5
        x = torch.where(flip.view(-1, 1, 1, 1), x.flip(3), x)
    if strong and COLOR_AUG:
        x = color_aug(x)
    x = augment(normalize(x))
    yb, lam = y, 1.0
    if strong and np.random.rand() < CUTMIX_P:
        x, y, yb, lam = cutmix(x, y)
    return x.contiguous(memory_format=torch.channels_last), y, yb, lam
```

```python
def build(k: int):
    torch.manual_seed(SEED + 1000 * k)           # different init per model
    d = DEVICES[k]
    m = VGG13(num_classes=NUM_CLASSES).to(d).to(memory_format=torch.channels_last)
    opt = optim.SGD(m.parameters(), lr=MAX_LR / 25, momentum=0.95, nesterov=True,
                    weight_decay=WD, fused=d.type == "cuda")
    return m, opt, GradScaler(d.type, enabled=enable_half)

models, opts, scalers = map(list, zip(*[build(k) for k in range(N_MODELS)]))
criterion = nn.CrossEntropyLoss(label_smoothing=LABEL_SMOOTH)

def one_cycle(p: float, max_lr=MAX_LR, pct=PCT_START, div=25.0, final_div=1e4, m_lo=0.85, m_hi=0.95):
    lo, end = max_lr / div, max_lr / div / final_div
    if p < pct:
        t = (1 - math.cos(math.pi * p / pct)) / 2
        return lo + (max_lr - lo) * t, m_hi + (m_lo - m_hi) * t
    t = (1 - math.cos(math.pi * min(1.0, (p - pct) / (1 - pct)))) / 2
    return max_lr + (end - max_lr) * t, m_lo + (m_hi - m_lo) * t

print(f"{N_MODELS} x VGG13, {sum(p.numel() for p in models[0].parameters()) / 1e6:.2f}M params each")
```

```python
def shift(x: Tensor, dx: int, dy: int) -> Tensor:
    H, W = x.shape[-2:]
    xp = F.pad(x, (1, 1, 1, 1), mode="replicate")
    return xp[:, :, 1 + dy:1 + dy + H, 1 + dx:1 + dx + W]

def _shifts(x):
    return [shift(x, 1, 0), shift(x, -1, 0), shift(x, 0, 1), shift(x, 0, -1)]

TTA_VIEWS = {
    "none":       lambda x: [x],
    "flip":       lambda x: [x, x.flip(3)],
    "flip+shift": lambda x: [x, x.flip(3)] + _shifts(x),
    "shift":      lambda x: [x] + _shifts(x),
}

@torch.inference_mode()
def predict_probs(model, X: Tensor, tta: str = "none", bs: int = 1024) -> Tensor:
    model.eval()
    d = next(model.parameters()).device
    outs = []
    for i in range(0, len(X), bs):
        x = prep(X[i:i + bs].to(d))
        views = TTA_VIEWS[tta](x)
        p = 0
        for v in views:
            with torch.autocast(d.type, enabled=enable_half):
                p = p + model(v.contiguous(memory_format=torch.channels_last)).float().softmax(1)
        outs.append((p / len(views)).cpu())
    model.train()
    return torch.cat(outs)

yv = Yva.cpu()
acc = lambda probs: 100.0 * (probs.argmax(1) == yv).float().mean().item()
```

```python
TRAIN_SECONDS = TRAIN_MINUTES * 60
n_tr = len(Xtr)
steps_per_epoch = n_tr // BS
has_val = len(Xva) > 0

def batch_indices():
    while True:
        perm = torch.randperm(n_tr, device=dev0)
        for i in range(steps_per_epoch):
            yield perm[i * BS:(i + 1) * BS]

idx_stream = batch_indices()
history, loss_sums, since_log = [], [torch.zeros((), device=d) for d in DEVICES], 0
print(f"{steps_per_epoch} steps/epoch, training for {TRAIN_MINUTES} min")
t_start = time.time()
step = 0
while True:
    p = (time.time() - t_start) / TRAIN_SECONDS
    if p >= 1.0:
        break
    lr, mom = one_cycle(p)
    kd_w = KD_WEIGHT * min(1.0, p / KD_RAMP) if N_MODELS > 1 else 0.0

    x, y, yb, lam = make_batch(next(idx_stream), strong=p < 1.0 - CLEAN_TAIL)
    batch = [(x.to(d), y.to(d), yb.to(d)) for d in DEVICES]

    logits = []
    for m, (xd, _, _) in zip(models, batch):
        with torch.autocast(xd.device.type, enabled=enable_half):
            z = m(xd)
        logits.append(z.float())
    if kd_w > 0:   
        soft = [F.softmax(z.detach() / KD_T, 1) for z in logits]
        targets = [sum(soft[j].to(DEVICES[k]) for j in range(N_MODELS) if j != k) / (N_MODELS - 1)
                   for k in range(N_MODELS)]

    for k, (m, opt, sc, z, (_, yd, ybd)) in enumerate(zip(models, opts, scalers, logits, batch)):
        loss = criterion(z, yd) if lam == 1.0 else lam * criterion(z, yd) + (1 - lam) * criterion(z, ybd)
        if kd_w > 0:
            loss = loss + kd_w * KD_T ** 2 * F.kl_div(F.log_softmax(z / KD_T, 1), targets[k],
                                                      reduction="batchmean")
        for g in opt.param_groups:
            g["lr"], g["momentum"] = lr, mom
        sc.scale(loss).backward()
        sc.step(opt)
        sc.update()
        opt.zero_grad(set_to_none=True)
        loss_sums[k] += loss.detach()
    step += 1
    since_log += 1

    if step % steps_per_epoch == 0:
        epoch = step // steps_per_epoch
        if epoch == 1 or epoch % EVAL_EVERY == 0:
            row = {"epoch": epoch, "minutes": round((time.time() - t_start) / 60, 2)}
            msg = f"epoch {epoch:4d} | {row['minutes']:5.1f} min"
            if has_val:
                row["val_ensemble"] = round(acc(sum(predict_probs(m, Xva) for m in models)), 2)
                msg += f" | accuracy {row['val_ensemble']:.2f}%"
            history.append(row)
            print(msg, flush=True)

for d in set(DEVICES):
    if d.type == "cuda":
        torch.cuda.synchronize(d)
train_minutes = (time.time() - t_start) / 60
print(f"training done: {step / steps_per_epoch:.0f} epochs in {train_minutes:.1f} min")
```

```python
BALANCE = True   # final run: True (it won on validation). Experiments with VAL_FRAC = 0.05: "auto"

def balance(P, iters=100):
    # make every class get the same share of predictions (train has exactly 500 per class)
    P = P.double().clone()
    target = P.shape[0] / P.shape[1]
    for _ in range(iters):
        P *= target / P.sum(0, keepdim=True)
        P /= P.sum(1, keepdim=True)
    return P.float()

tta_opts = ["none", "flip", "flip+shift"] if USE_FLIP else ["none", "shift"]
use_balance = BALANCE is True or (BALANCE == "auto" and not has_val)
tta = tta_opts[-1]
if has_val:
    res = {}
    for t in tta_opts:
        p = sum(predict_probs(m, Xva, t) for m in models)
        res[t] = (acc(p), acc(balance(p)))
    tta = max(tta_opts, key=lambda t: res[t][0])
    if BALANCE == "auto":
        use_balance = res[tta][1] > res[tta][0]
    print(f"final accuracy: {res[tta][1 if use_balance else 0]:.2f}%")

P = sum(predict_probs(m, Xte, tta) for m in models)
pred = (balance(P) if use_balance else P).argmax(1).numpy()

out_dir = "/kaggle/working" if os.path.exists("/kaggle/working") else "."
for f in os.listdir(out_dir):                      # keep ONLY submission.csv in the output
    if f.endswith((".csv", ".json", ".pt")):
        os.remove(os.path.join(out_dir, f))
pd.DataFrame({"ID": np.arange(len(pred)), "target": pred}).to_csv(os.path.join(out_dir, "submission.csv"), index=False)
print(f"total time: {(time.time() - NOTEBOOK_T0) / 60:.1f} min")
```
